```python
def analyze(marks):
    total = sum(marks)
    avg = total / len(marks)

    if avg > 90:
        print("excellent")
    elif avg > 80:
        print("good")
    elif avg > 70:
        print("well")
    else:
        print("need improvement")

    for i in range(0, 4):
        print(marks[i])

    print("Total:", total)
    print("Average:", round(avg, 2))


analyze([95, 85, 75, 90])
```
